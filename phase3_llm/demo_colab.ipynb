## 🚀 Load LLM and Initialize System

**Note:** Make sure you have a HuggingFace token with Llama access set in Colab Secrets as `HF_TOKEN`


```python
# 4️⃣ Load Llama model (takes 2-3 minutes with 4-bit quantization)
from phase3_llm.llm_wrapper import LLMWrapper

print("⏳ Loading Llama-3.1-8B-Instruct (4-bit)...")
llm = LLMWrapper(model_size="8b")
llm.load()
print("✓ LLM ready!")
```

```python
# 5️⃣ Initialize Research Assistant (re-run after code updates)
from phase3_llm.pipeline import ResearchAssistant

assistant = ResearchAssistant(
    llm=llm,
    db_path="phase1_data_preparation/outputs/papers.db",
    index_path="phase1_data_preparation/outputs/faiss_index.bin",
    metadata_path="phase1_data_preparation/outputs/chunk_metadata.json"
)
print("✅ Research Assistant ready! Let's query some papers!")
```

## 💡 Demo Queries

The system automatically routes queries to the right tools!

**📋 Tool Types:**
- **Database Tool** → Structured queries (datasets, models, training setups)
- **Vector Search (RAG)** → Semantic search of paper content
- **Web Search** → arXiv API for external papers


```python
# 🗄️ Query 1: Database query
result = assistant.answer("What are the most common evaluation datasets?")
```

```python
# 🔍 Query 2: Semantic search
result = assistant.answer("Tell me about OpenVLA")
```

```python
# 🌐 Query 3: Web search (arXiv)
result = assistant.answer("Look up recent papers on sim-to-real robot learning")
```

```python
# 🔍 Query 4: Conceptual question (semantic search)
result = assistant.answer("How does sim-to-real transfer work in robot learning?")
```

```python
# 🗄️ Query 5: Database query (vision models)
result = assistant.answer("What vision models are used across papers?")
```

```python
# 🔍 Query 6: Multi-tool (semantic + database)
result = assistant.answer("Compare vision-language models in manipulation tasks")
```
